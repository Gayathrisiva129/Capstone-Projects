##Task 2 — Data Cleaning
###●  Cleaning Missing Data
###■ Detect missing values (isnull(), isna())
###■ Handle using: dropna(), fillna(), median/mode, forward/backward-fill

```python
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
df = df.dropna()
df.head()
```

```python
df.shape
```

```python
df["age"] = df["age"].fillna(df["age"].median())
df.head()
```

```python
df.isnull().sum()
```

```python
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.ffill()
df.head()
```

```python
df = df.bfill()
df.head()
```

```python
df.isnull() .sum()
```

###Duplicate Handling- Identify & remove duplicates

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.head()
```

```python
df.duplicated() .sum()
df. head()
```

##Data Formatting
###■ Fix incorrect data types
###■ Standardize categories (e.g., gender labels)
###■ Correct unrealistic values in likes, comments, shares

```python
df.dtypes
```

```python
df["gender"] = df["gender"].str.lower().str.strip()
df.head()
```

```python
df.loc[df["likes"] < 0, "likes"] = 0
df.loc[df["comments"] < 0, "comments"] = 0
df.loc[df["shares"] < 0, "shares"] = 0
df.head(5000)
```

##●  Feature Cleaning
###■ Extract hashtag count
###■ Clean sentiment labels

```python
df["hashtag_count"] = df["hashtags"].str.count("#")
df.head(5000)
```

```python
df["sentiment"] = df["sentiment"].str.lower().str.strip()
df.head(5000)
```

##❖ Task 6 — Data Visualization (Min. 8 Plots Required)
###● Matplotlib
####○ Scatter: likes vs impressions

```python
import matplotlib.pyplot as pyplt
```

```python
pyplt.scatter(df["likes"], df["impression_count"])
pyplt.xlabel("Likes")
pyplt.ylabel("Impressions")
pyplt.title("Likes vs Impressions")
pyplt.show()
```

####○ Line: daily engagement trend

```python
df.groupby("posted_at")["engagement_rate"].mean().plot()
pyplt.title("Daily Engagement Trend")
pyplt.xlabel("Date")
pyplt.ylabel("Engagement Rate")
pyplt.show()
```

####○ Bar: posts by category

```python
df["post_category"].value_counts().plot(kind="bar")
pyplt.title("Posts by Category")
pyplt.xlabel("Category")
pyplt.ylabel("Number of Posts")
pyplt.show()
```

####○ Pie: gender distribution

```python
df["gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")
pyplt.title("Gender Distribution")
pyplt.ylabel("")
pyplt.show()
```

####○ Histogram: age

```python
pyplt.hist(df["age"])
pyplt.title("Age Distribution")
pyplt.xlabel("Age")
pyplt.ylabel("Count")
pyplt.show()
```

####○ Box: engagement rate

```python
pyplt.boxplot(df["engagement_rate"])
pyplt.title("Engagement Rate")
pyplt.ylabel("Engagement Rate")
pyplt.show()
```

##Seaborn
###○ Count plot: post type

```python
import seaborn as sbn
sbn.countplot(x="post_type", data=df)
pyplt.title("Post Type Count")
pyplt.xlabel("Post Type")
pyplt.ylabel("Count")
pyplt.show()
```

###○ Bar plot: avg likes by category

```python
sbn.barplot(data=df, x="post_category", y="likes")
pyplt.title("Average Likes by Category")
pyplt.xlabel("Post Category")
pyplt.ylabel("Average Likes")
pyplt.show()
```

###○ Violin: followers vs sentiment

```python
sbn.violinplot(data=df, x="sentiment", y="follower_count")
pyplt.title("Followers vs Sentiment")
pyplt.xlabel("Sentiment")
pyplt.ylabel("Followers")
pyplt.show()
```

###○ Pair plot: numeric features

```python
sbn.pairplot(df[["likes", "comments", "shares", "age"]])
pyplt.show()
```

###○ Heatmap: correlation matrix

```python
sbn.heatmap(df.corr(numeric_only=True), annot=True)
pyplt.title("Correlation Heatmap")
pyplt.show()
```

###○ Swarm plot: engagement vs device

```python
sbn.swarmplot(data=df.sample(100), x="device_type", y="engagement_rate")
pyplt.title("Engagement vs Device")
pyplt.xlabel("Device Type")
pyplt.ylabel("Engagement Rate")
pyplt.show()
```
